```python
# ===== PYTHON OPERATORS  =====

a = 10
b = 3

# --- Arithmetic Operators ---
print("--- Arithmetic ---")
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)

# --- Comparison Operators ---
print("--- Comparison ---")
print("a == b:", a == b)
print("a != b:", a != b)
print("a > b:", a > b)
print("a < b:", a < b)
print("a >= b:", a >= b)
print("a <= b:", a <= b)

# --- Assignment Operators ---
print("--- Assignment ---")
x = 5
print("x =", x)
x += 3
print("x += 3 ->", x)
x -= 2
print("x -= 2 ->", x)
x *= 4
print("x *= 4 ->", x)
x /= 6
print("x /= 6 ->", x)

# --- Logical Operators ---
print("--- Logical ---")
p = True
q = False
print("p and q:", p and q)
print("p or q:", p or q)
print("not p:", not p)

# --- Membership Operators ---
print("--- Membership ---")
fruits = ["apple", "banana", "cherry"]
print("'apple' in fruits:", "apple" in fruits)
print("'mango' not in fruits:", "mango" not in fruits)

# --- Identity Operators ---
print("--- Identity ---")
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1
print("list1 is list3:", list1 is list3)
print("list1 is list2:", list1 is list2)
print("list1 == list2:", list1 == list2)

# --- Bitwise Operators ---
print("--- Bitwise ---")
m = 5   # binary 0101
n = 3   # binary 0011
print("m & n:", m & n)
print("m | n:", m | n)
print("m ^ n:", m ^ n)
print("m << 1:", m << 1)
print("m >> 1:", m >> 1)
```

```python
# ===== PYTHON CONDITIONS  =====

# --- if statement ---
age = 18
if age >= 18:
    print("You are an adult")

# --- if / else ---
number = 7
if number % 2 == 0:
    print(number, "is even")
else:
    print(number, "is odd")

# --- if / elif / else ---
marks = 85
if marks >= 90:
    print("Grade A+")
elif marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("Fail")

# --- Using 'and' / 'or' ---
age = 20
has_id = True
if age >= 18 and has_id:
    print("Entry allowed")
else:
    print("Entry denied")

day = "Sunday"
if day == "Saturday" or day == "Sunday":
    print("It's the weekend")

# --- Nested if ---
num = 15
if num > 0:
    print("Positive number")
    if num > 10:
        print("It is also greater than 10")

# --- Check positive, negative or zero ---
value = -4
if value > 0:
    print("Positive")
elif value < 0:
    print("Negative")
else:
    print("Zero")

# --- Short if (ternary) ---
x = 10
result = "Big" if x > 5 else "Small"
print(result)

# --- Check if item is in a list ---
colors = ["red", "green", "blue"]
if "red" in colors:
    print("Red is in the list")
```

```python
# ===== PYTHON LOOPS  =====

# --- for loop with range ---
print("--- for loop: 1 to 5 ---")
for i in range(1, 6):
    print(i)

# --- range with step ---
print("--- Even numbers 2 to 10 ---")
for i in range(2, 11, 2):
    print(i)

# --- Loop through a list ---
print("--- Loop through list ---")
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)

# --- Loop through a string ---
print("--- Loop through string ---")
for letter in "Python":
    print(letter)

# --- while loop ---
print("--- while loop ---")
count = 1
while count <= 5:
    print("Count:", count)
    count += 1

# --- break ---
print("--- break ---")
for i in range(1, 10):
    if i == 5:
        break
    print(i)

# --- continue ---
print("--- continue (skip 3) ---")
for i in range(1, 6):
    if i == 3:
        continue
    print(i)

# --- Sum of numbers 1 to 10 ---
total = 0
for i in range(1, 11):
    total += i
print("Sum 1 to 10 =", total)

# --- Multiplication table ---
print("--- Table of 5 ---")
for i in range(1, 11):
    print("5 x", i, "=", 5 * i)

# --- Nested loop (star pattern) ---
print("--- Star pattern ---")
for i in range(1, 5):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
# ===== PYTHON FUNCTIONS (Beginner) =====

# --- Simple function ---
def say_hello():
    print("Hello, World!")

say_hello()

# --- Function with a parameter ---
def greet(name):
    print("Hello,", name)

greet("Ali")
greet("Sara")

# --- Function with return value ---
def add(a, b):
    return a + b

result = add(5, 3)
print("5 + 3 =", result)

# --- Function with default parameter ---
def welcome(name="Guest"):
    print("Welcome,", name)

welcome()
welcome("Ahmed")

# --- Even or odd ---
def check_even_odd(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"

print("4 is", check_even_odd(4))
print("7 is", check_even_odd(7))

# --- Simple calculator functions ---
def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b

print("10 - 4 =", subtract(10, 4))
print("10 * 4 =", multiply(10, 4))
print("10 / 4 =", divide(10, 4))
print("10 / 0 =", divide(10, 0))

# --- Function using a loop ---
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print("Factorial of 5 =", factorial(5))

# --- Function returning multiple values ---
def min_max(numbers):
    return min(numbers), max(numbers)

smallest, largest = min_max([4, 9, 1, 7])
print("Smallest:", smallest, "Largest:", largest)

# --- Lambda (short function) ---
square = lambda x: x * x
print("Square of 6 =", square(6))
```

```python
# ===== PYTHON ARRAYS / LISTS  =====
# In Python, lists are used as arrays.

# --- Create a list ---
numbers = [10, 20, 30, 40, 50]
print("List:", numbers)

# --- Access items (index starts at 0) ---
print("First item:", numbers[0])
print("Third item:", numbers[2])
print("Last item:", numbers[-1])

# --- Change an item ---
numbers[1] = 25
print("After change:", numbers)

# --- Add items ---
numbers.append(60)          # add at the end
print("After append:", numbers)

numbers.insert(0, 5)        # add at a position
print("After insert:", numbers)

# --- Remove items ---
numbers.remove(25)          # remove by value
print("After remove:", numbers)

numbers.pop()               # remove last item
print("After pop:", numbers)

# --- Length of list ---
print("Length:", len(numbers))

# --- Loop through list ---
print("--- Loop ---")
for n in numbers:
    print(n)

# --- Slicing ---
print("First 3 items:", numbers[:3])
print("From index 2:", numbers[2:])
print("Index 1 to 3:", numbers[1:4])

# --- Sorting and reversing ---
data = [4, 2, 9, 1, 7]
data.sort()
print("Sorted:", data)
data.reverse()
print("Reversed:", data)

# --- Useful functions ---
print("Sum:", sum(data))
print("Min:", min(data))
print("Max:", max(data))

# --- Check if item exists ---
if 9 in data:
    print("9 is in the list")

# --- Sum using loop ---
total = 0
for n in data:
    total += n
print("Total using loop:", total)

# --- List of strings ---
fruits = ["apple", "banana", "cherry"]
fruits.append("mango")
print("Fruits:", fruits)
print("Number of fruits:", len(fruits))

# --- 2D list (list inside list) ---
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print("Matrix:", matrix)
print("Row 2, Column 3:", matrix[1][2])

for row in matrix:
    print(row)
```
